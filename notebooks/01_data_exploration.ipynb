Setting up X, y, and groups

```python
X = df.drop(columns =["name", "session_id","subject_id","status","Jitter:DDP", "Shimmer:DDA"])
y = df["status"]

groups = df["subject_id"]
```

```python
#Verifying 
print("X shape:", X.shape)
print("y length:", len(y))
print("groups length:", len(groups))
print("unique subjects:", groups.nunique())
```

StratifiedGroupKFold

```python
from sklearn.model_selection import StratifiedGroupKFold, LeaveOneGroupOut
```

```python
# Creating CrossValidator
sgkf = StratifiedGroupKFold ( 
    n_splits = 5, 
    shuffle = True, 
    random_state = 42 ) 
```

```python
for fold, (train_idx, test_idx) in enumerate(
    sgkf.split(X, y, groups),
    start=1
):
    test_subject_labels = (
        df.iloc[test_idx]
        .groupby("subject_id")["status"]
        .first()
    )

    healthy_count = (test_subject_labels == 0).sum()
    pd_count = (test_subject_labels == 1).sum()

    print(f"Fold {fold}: Healthy={healthy_count}, PD={pd_count}")
```

```python
logo = LeaveOneGroupOut()

print("Number of folds:", logo.get_n_splits(X,y,groups))

for fold, (train_idx, test_idx) in enumerate ( 
    logo.split(X,y,groups) , 
    start = 1 
):
    test_subject_labels =(
    df.iloc[test_idx].groupby("subject_id")["status"].first() 
    ) 

    healthy_count = (test_subject_labels == 0).sum()
    pd_count = (test_subject_labels == 1).sum()

    print(f"Fold {fold}: Healthy={healthy_count}, PD={pd_count}")
        
```

Building the preprocessing pipeline 

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PowerTransformer, RobustScaler 
from sklearn.compose import ColumnTransformer

skewed_cols  = skewed_features.index.tolist()
all_cols = feature_cols.columns.tolist()
```

```python
# Building Column transformer 
# Column transformer is used to apply multiple transformation
preprocessor = ColumnTransformer (
    transformers = [
        (
            "power_transform", 
            PowerTransformer(
                method = "yeo-johnson", 
                standardize = False
            ), 
            skewed_cols
        ) ,
    ],
        

        remainder = "passthrough"
        
    
)

# Building Pipeline 

full_pipeline = Pipeline (
    [
        ("preprocessing", preprocessor),
        ("scaling", RobustScaler()),
    ]
)
```

```python
# testing on one fold manually to check if PowerTransform is 
# actually working 

train_idx, test_idx = next(sgkf.split(X,y,groups))

```

```python
X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]
```

```python
# Fitting the pipeline with only training data 
X_train_transformed = full_pipeline.fit_transform(X_train)
```

```python

X_test_transformed = full_pipeline.transform(X_test)
```

```python
#checking shapes 
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("Transformed train:", X_train_transformed.shape)
print("Transformed test:", X_test_transformed.shape)
```

```python
# Columns transformer will first produce output for 12 skewed data 
# and then the rest 8 data. so we need that order to be maintained 

correct_columns = full_pipeline.named_steps["preprocessing"].get_feature_names_out()
print(correct_columns)
```

```python
# correct_columns gives the output of the Columns Transformer so 

X_train_transformed_df = pd.DataFrame (
    X_train_transformed, 
    columns = correct_columns, 
    index = X_train.index

)
print(X_train_transformed_df.shape)
```

Creating a baseline using DummyClassifier
mostfrequent and stratifiend

```python
from sklearn.dummy import DummyClassifier
```

```python
#creating a place to store the prediction for mostFrea

oof_pred_most_frequent = np.empty(len(y) , dtype = int) 
```

```python
# applying most_frequent for all the 5 folds from skgf 

for fold, (train_idx, test_idx ) in enumerate ( 
     sgkf.split(X,y,groups), 
     start = 1 
 ): 
     X_train = X.iloc[train_idx]
     X_test = X.iloc[test_idx] 

     y_train = y.iloc[train_idx] 

     dummy = DummyClassifier ( 
             strategy = "most_frequent"
     ) 

     dummy.fit(X_train,y_train)

     fold_pred = dummy.predict(X_test) 

     oof_pred_most_frequent[test_idx] = fold_pred 

     print(
        f"Fold {fold}: "
        f"Train={len(train_idx)}, "
        f"Test={len(test_idx)}"
    )
```

```python
# evaluating most_frequent 
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(
    classification_report(
        y, # actual values 
        oof_pred_most_frequent, # preficted values 
        target_names=["Healthy", "PD"], # 0 and 1 respectively 
        zero_division=0 
        
    )

    # whenever 0 divsion return 0 instead of a warning 
    # so zero_division = 0 
)



```

```python
# confusion metric 

print (confusion_matrix (
    y, 
    oof_pred_most_frequent
))
```

```python
# now for stratified 

oof_pred_stratified = np.empty(len(y), dtype=int)
```

```python
for fold, (train_idx, test_idx) in enumerate(
    sgkf.split(X, y, groups),
    start=1
):
    X_train = X.iloc[train_idx]
    X_test = X.iloc[test_idx]

    y_train = y.iloc[train_idx]

    dummy = DummyClassifier(
        strategy="stratified",
        random_state=42
    )

    dummy.fit(X_train, y_train)

    fold_pred = dummy.predict(X_test)

    oof_pred_stratified[test_idx] = fold_pred
```

```python
print(
    classification_report(
        y,
        oof_pred_stratified,
        target_names=["Healthy", "PD"]
    )
)
```

```python
print(
    confusion_matrix(
        y,
        oof_pred_stratified
    )
)
```
